# 2. Import Library

```python
import os
import re
import nltk
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
# Lightweight NLTK parsing components
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.sentiment import SentimentIntensityAnalyzer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
# Hugging Face extensions for genuine DistilBERT Training
from datasets import Dataset as HFDataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
# Download required low-overhead NLTK resource structures
nltk.download('vader_lexicon', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
sns.set_theme(style='whitegrid')
def plot_academic_confusion_matrix(y_true, y_pred, model_title):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='mako',
                xticklabels=['Negative', 'Positive'],
                yticklabels=['Negative', 'Positive'])
    plt.title(f'{model_title} Confusion Matrix')
    plt.ylabel('True Target Label')
    plt.xlabel('Predicted Label Paradigm')
    plt.show()
```

# 3. Download and Load the Data

```python
print("Downloading and retrieving full 50,000 records of IMDB Dataset via Kagglehub...")
cache_path = kagglehub.dataset_download("lakshmi25npathi/imdb-dataset-of-50k-movie-reviews")
csv_file_path = os.path.join(cache_path, "IMDB Dataset.csv")
df = pd.read_csv(csv_file_path)
print(f"Dataset successfully loaded from Kaggle cache! Matrix dimensions: {df.shape}")
df['label'] = df['sentiment'].map({'positive': 1, 'negative': 0})
# Visualizing Pure Binary Class Balance (Assigned hue parameter to suppress warning)
plt.figure(figsize=(10, 6))
sns.countplot(x='sentiment', hue='sentiment', data=df, palette='mako', legend=False)
plt.title('Dataset Class Distribution (Strict 50K Binary Balance Check)')
plt.show()
```

# 4. Data Preprocessing

```python
lemmatizer = WordNetLemmatizer()
eng_stopwords = set(stopwords.words('english'))
def execute_preprocessing_pipeline(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r'<[^>]+>', ' ', text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = word_tokenize(text)
    cleaned_tokens = [lemmatizer.lemmatize(token) for token in tokens if token not in eng_stopwords and len(token) > 2]
    return " ".join(cleaned_tokens)
print("Running lightning-fast NLTK preprocessing checklist over full 50,000 records...")
df['cleaned_review'] = df['review'].apply(execute_preprocessing_pipeline)
# Fixed: Split the main DataFrame first to guarantee index alignment and avoid data leakage
df_temp, df_test = train_test_split(
    df, test_size=0.2, stratify=df['label'], random_state=42
)
df_train, df_val = train_test_split(
    df_temp, test_size=0.125, stratify=df_temp['label'], random_state=42
)
# Extract respective text splits
X_train = df_train['cleaned_review']
X_val = df_val['cleaned_review']
X_test = df_test['cleaned_review']
y_train = df_train['label']
y_val = df_val['label']
y_test = df_test['label']
X_train_raw = df_train['review']
X_val_raw = df_val['review']
X_test_raw = df_test['review']
print(f"\nData Three-Way Stratified Partition Finalized Successfully:")
print(f"-> [TRAIN SET Size]       : {X_train.shape[0]} reviews (70%)")
print(f"-> [VALIDATION SET Size]  : {X_val.shape[0]} reviews (10%)")
print(f"-> [TEST SET Size]        : {X_test.shape[0]} reviews (20%)")
```

# 4.1 Exploratory Data Analysis (EDA)

```python
# --- Review Length Distribution ---
df['review_length'] = df['cleaned_review'].apply(lambda x: len(x.split()))
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(data=df, x='review_length', hue='sentiment', bins=50, palette='mako', kde=True)
plt.title('Review Length Distribution by Sentiment')
plt.xlabel('Number of Words')
plt.ylabel('Count')
plt.xlim(0, 500)

# --- Word Clouds ---
plt.subplot(1, 2, 2)
positive_text = ' '.join(df[df['label']==1]['cleaned_review'].dropna())
wc_pos = WordCloud(width=800, height=400, background_color='white', colormap='viridis', max_words=100).generate(positive_text)
plt.imshow(wc_pos, interpolation='bilinear')
plt.axis('off')
plt.title('Positive Reviews Word Cloud')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 5))
negative_text = ' '.join(df[df['label']==0]['cleaned_review'].dropna())
wc_neg = WordCloud(width=800, height=400, background_color='white', colormap='inferno', max_words=100).generate(negative_text)
plt.subplot(1, 2, 1)
plt.imshow(wc_neg, interpolation='bilinear')
plt.axis('off')
plt.title('Negative Reviews Word Cloud')

# --- Sample Cleaned vs Raw Text ---
plt.subplot(1, 2, 2)
plt.axis('off')
sample_idx = df.index[0]
sample_text = f"Raw: {df.loc[sample_idx, 'review'][:150]}...\n\nCleaned: {df.loc[sample_idx, 'cleaned_review'][:150]}..."
plt.text(0.05, 0.5, sample_text, fontsize=9, verticalalignment='center', wrap=True,
         bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
plt.title('Sample: Raw vs Cleaned Review')
plt.tight_layout()
plt.show()

print(f"\nAverage review length (words): {df['review_length'].mean():.1f}")
print(f"Median review length (words): {df['review_length'].median():.1f}")
print(f"Max review length (words): {df['review_length'].max()}")
```

# 5. LEXICON-BASED (Vader & VADER Threshold Optimizing)

```python
print("\n--- 1. Lexicon-Based Era: VADER Threshold Calibration ---")
sia = SentimentIntensityAnalyzer()
best_vader_acc = 0
optimal_boundary = 0.05
threshold_candidates = np.arange(0.01, 0.51, 0.01)
# NOTE: VADER uses raw (uncleaned) text because it relies on punctuation, casing, and stopwords
# to compute accurate sentiment scores. Pre-processed text would lose these important signals.
print("Pre-computing VADER compound scores on Train Set (Once for performance)...")
# Fixed: Extract polarity scores once to prevent nested-loop lag
train_compound_scores = X_train_raw.apply(lambda x: sia.polarity_scores(x)['compound']).values
print("Calibrating VADER binary decision boundaries granularly on Train Set...")
for t in threshold_candidates:
    train_preds = (train_compound_scores >= t).astype(int)
    current_acc = accuracy_score(y_train, train_preds)
    if current_acc > best_vader_acc:
        best_vader_acc = current_acc
        optimal_boundary = t
print(f"VADER Calibration Complete. Chosen Optimal Boundary Threshold: {optimal_boundary:.2f}")
# Precompute and test VADER model
test_compound_scores = X_test_raw.apply(lambda x: sia.polarity_scores(x)['compound']).values
final_vader_preds = (test_compound_scores >= optimal_boundary).astype(int)
vader_test_acc = accuracy_score(y_test, final_vader_preds)
vader_test_f1 = f1_score(y_test, final_vader_preds, average='macro')
print(f"VADER Final Test Accuracy: {vader_test_acc:.4f} | Macro F1: {vader_test_f1:.4f}")
plot_academic_confusion_matrix(y_test, final_vader_preds, "VADER Baseline")
```

# 6. TRADITIONAL ML (Logistic Regression + GridSearch)


```python
print("\n--- 2. Traditional ML Era: Logistic Regression Training with Tuning ---")
tfidf = TfidfVectorizer(max_features=10000, ngram_range=(1, 2))
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
lr_param_grid = {
    'C': [0.1, 1.0, 10.0],
    'solver': ['liblinear', 'lbfgs']
}
base_lr = LogisticRegression(max_iter=500, random_state=42)
print("Launching GridSearchCV coupled with 5-Fold Stratified Cross-Validation...")
grid_search_lr = GridSearchCV(estimator=base_lr, param_grid=lr_param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search_lr.fit(X_train_tfidf, y_train)
best_lr = grid_search_lr.best_estimator_
print(f"Optimal Logistic Regression Parameters Identified: {grid_search_lr.best_params_}")
lr_preds = best_lr.predict(X_test_tfidf)
lr_acc = accuracy_score(y_test, lr_preds)
lr_f1 = f1_score(y_test, lr_preds, average='macro')
print(f"Tuned Logistic Regression Test Accuracy: {lr_acc:.4f} | Macro F1: {lr_f1:.4f}")
print(classification_report(y_test, lr_preds, target_names=['Negative', 'Positive']))
plot_academic_confusion_matrix(y_test, lr_preds, "Tuned LogReg")
```

# 7. DEEP LEARNING (LSTM Neural Network Model Training)

```python
print("\n--- 3. Deep Learning Era: LSTM Neural Network Model Training ---")
MAX_VOCAB = 12000
MAX_LEN = 200
tokenizer = Tokenizer(num_words=MAX_VOCAB, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)
X_train_seq = pad_sequences(tokenizer.texts_to_sequences(X_train), maxlen=MAX_LEN, padding='post')
X_val_seq = pad_sequences(tokenizer.texts_to_sequences(X_val), maxlen=MAX_LEN, padding='post')
X_test_seq = pad_sequences(tokenizer.texts_to_sequences(X_test), maxlen=MAX_LEN, padding='post')
# Fixed recurrent network architecture template with regularization design layers
lstm_model = Sequential([
    Embedding(MAX_VOCAB, 128, input_length=MAX_LEN),
    LSTM(64, return_sequences=False),
    Dropout(0.5),                        # Structural parameter regularizer to suppress overtraining constraints
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
lstm_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])
early_stop_callback = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)
print("Training Deep Neural LSTM Network with empirical configurations and optimization...")
history = lstm_model.fit(
    X_train_seq, y_train, epochs=10, batch_size=64,
    validation_data=(X_val_seq, y_val), callbacks=[early_stop_callback], verbose=1
)
lstm_probs = lstm_model.predict(X_test_seq, verbose=0)
lstm_preds = (lstm_probs >= 0.5).astype(int).flatten()
lstm_acc = accuracy_score(y_test, lstm_preds)
lstm_f1 = f1_score(y_test, lstm_preds, average='macro')
print(f"LSTM Deep Learning Model Test Accuracy: {lstm_acc:.4f} | Macro F1: {lstm_f1:.4f}")
print(classification_report(y_test, lstm_preds, target_names=['Negative', 'Positive']))
plot_academic_confusion_matrix(y_test, lstm_preds, "LSTM Network")
# Overfitting Check Trajectory Visualizer
plt.figure(figsize=(10, 4))
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Deep LSTM Training Trajectory Distribution')
plt.ylabel('Accuracy Boundary')
plt.xlabel('Epochs')
plt.legend()
plt.show()
```

# 8. TRANSFORMER ERA (GENUINE DISTILBERT MODEL FINE-TUNING)

```python
print("\n--- 4. Transformer Era: Transformer Architecture (DistilBERT) Fine-Tuning ---")
# Fixed: Feed raw text partitions (X_train_raw, etc.) to the Transformer.
# Pre-trained models like DistilBERT require punctuation, casing, and stopwords to process semantics correctly.
train_hf = HFDataset.from_dict({"text": X_train_raw.tolist(), "label": y_train.tolist()})
val_hf = HFDataset.from_dict({"text": X_val_raw.tolist(), "label": y_val.tolist()})
test_hf = HFDataset.from_dict({"text": X_test_raw.tolist(), "label": y_test.tolist()})
model_ckpt = "distilbert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(model_ckpt)
def tokenize_fn_full(examples):
    return bert_tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)
print("Tokenizing split dataset string structures for DistilBERT arrays...")
tokenized_train = train_hf.map(tokenize_fn_full, batched=True)
tokenized_val = val_hf.map(tokenize_fn_full, batched=True)
tokenized_test = test_hf.map(tokenize_fn_full, batched=True)
# Fixed: Updated training argument 'evaluation_strategy' to modern 'eval_strategy'
train_args_full = TrainingArguments(
    output_dir="./distilbert_full_split_output",
    eval_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    gradient_accumulation_steps=4,     # Emulated batch parameters: 16 * 4 = 64
    num_train_epochs=2,
    weight_decay=0.01,
    logging_steps=200,
    report_to="none"
)
bert_model_full = AutoModelForSequenceClassification.from_pretrained(model_ckpt, num_labels=2)
trainer_full = Trainer(
    model=bert_model_full,
    args=train_args_full,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val
)
print("Executing genuine DistilBERT fine-tuning via active GPU hardware backpropagation...")
trainer_full.train()
print("Evaluating fine-tuned DistilBERT model weights across the test subset...")
raw_predictions_full = trainer_full.predict(tokenized_test)
bert_preds = np.argmax(raw_predictions_full.predictions, axis=-1)
bert_labels = y_test.tolist()
bert_acc = accuracy_score(bert_labels, bert_preds)
bert_f1 = f1_score(bert_labels, bert_preds, average='macro')
print(f"Fine-Tuned DistilBERT Test Partition Accuracy: {bert_acc:.4f} | Macro F1: {bert_f1:.4f}")
print(classification_report(bert_labels, bert_preds, target_names=['Negative', 'Positive']))
plot_academic_confusion_matrix(bert_labels, bert_preds, "Fine-Tuned DistilBERT")
```

# 9. SYSTEM ARCHITECTURAL COMPARATIVE VISUALIZATION (MULTI-METRIC)

```python
architecture_paradigms = ['VADER Baseline', 'Tuned LogReg', 'LSTM Network', 'Fine-Tuned DistilBERT']
accuracies_list = [vader_test_acc, lr_acc, lstm_acc, bert_acc]
f1_scores_list = [vader_test_f1, lr_f1, lstm_f1, bert_f1]
plot_df = pd.DataFrame({
    'Model Paradigm': architecture_paradigms * 2,
    'Observed Score': accuracies_list + f1_scores_list,
    'Metric Type': ['Accuracy'] * 4 + ['Macro F1-Score'] * 4
})
plt.figure(figsize=(11, 6))
bar_plot = sns.barplot(x='Model Paradigm', y='Observed Score', hue='Metric Type', data=plot_df, palette='mako')
plt.title('Cross-Era Architectural Performance Benchmarking (Strict 3-Way Split Alignment)', fontsize=12, fontweight='bold')
plt.ylabel('Observed Test Performance')
plt.ylim(0.5, 1.0)
for bar in bar_plot.patches:
    y_val = bar.get_height()
    if y_val > 0:
        plt.text(bar.get_x() + bar.get_width()/2.0, y_val + 0.01, f"{y_val:.4f}", ha='center', va='bottom', fontsize=9, fontweight='bold')
plt.legend(loc='lower right')
plt.show()
```

# 10. LEXICON-BASED ASPECT SENTIMENT DEMONSTRATION

```python
print("\n--- 5. Lexical Exploration: Dependency-Based Opinion Mining ---")
import spacy
# Load spaCy English model (Pre-installed in Google Colab)
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    # Download the package if it's missing
    import subprocess
    import sys
    subprocess.run([sys.executable, "-m", "spacy", "download", "en_core_web_sm"])
    nlp = spacy.load("en_core_web_sm")
def execute_dependency_opinion_mining(review_string):
    print(f"\nIngested Critique Text: \"{review_string}\"")
    print(f"{'Opinion Target (Noun)':<25} | {'Opinion Modifier':<25} | {'Dependency Relation':<25} | {'Polarity'}")
    print("-" * 100)

    doc = nlp(review_string)

    for token in doc:
        # 1. Identify Opinion Targets (Noun or Proper Noun)
        if token.pos_ in ["NOUN", "PROPN"]:

            # Relation A: Direct Adjectival Modifiers (amod)
            # Example: "stellar cinematography" -> Target: cinematography, Modifier: stellar
            for child in token.children:
                if child.dep_ == "amod" and child.pos_ in ["ADJ", "VERB"]:
                    score = sia.polarity_scores(child.text)['compound']
                    if abs(score) > 0.05:
                        polarity = "POSITIVE" if score > 0 else "NEGATIVE"
                        print(f"{token.text:<25} | {child.text:<25} | {f'amod ({child.dep_})':<25} | {polarity}")

            # Relation B: Predicate Adjectives (nsubj -> copula verb -> adjective complement)
            # Example: "the script was disappointing" -> Target: script, Modifier: disappointing
            if token.dep_ == "nsubj":
                verb = token.head
                for child in verb.children:
                    # Look for adjectival complements (acomp) or attribute descriptions
                    if child.dep_ in ["acomp", "oprd", "attr"] and child.pos_ in ["ADJ", "VERB"]:
                        score = sia.polarity_scores(child.text)['compound']
                        if abs(score) > 0.05:
                            polarity = "POSITIVE" if score > 0 else "NEGATIVE"
                            print(f"{token.text:<25} | {child.text:<25} | {f'nsubj-verb-{child.dep_}':<25} | {polarity}")
# Test the upgraded dependency parser
execute_dependency_opinion_mining("The stellar cinematography and fantastic acting salvaged the movie, though the lengthy script was deeply disappointing.")
```

# ABSA
1. import spacy as noun extraction

2. import fine-tuned transformer models-> for +ve and -ve cases only and lightweight distillbert

3. Extract Some Test Set to evaluate

4. UAsing the distillbert-ABSA to categorize the aspect

```python
import spacy
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. 环境初始化与轻量化 Transformer 模型加载
# ==========================================
try:
    nlp = spacy.load("en_core_web_sm", disable=["ner", "textcat"])
except OSError:
    import subprocess
    import sys
    subprocess.run([sys.executable, "-m", "spacy", "download", "en_core_web_sm"])
    nlp = spacy.load("en_core_web_sm", disable=["ner", "textcat"])

model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

movie_aspects = {
    'STORY': ['plot', 'story', 'script', 'screenplay', 'narrative'],
    'PERFORMANCE': ['acting', 'cast', 'actor', 'actress', 'performance'],
    'DIRECTION': ['direction', 'director', 'directing'],
    'VISUALS': ['cinematography', 'effects', 'cgi', 'visuals'],
    'MUSIC': ['music', 'soundtrack', 'score', 'sound']
}

# 情感预测核心推理函数 (严格对齐 SST-2 的二分类标签)
def predict_aspect_sentiment(sentence, aspect_term):
    inputs = tokenizer(sentence, aspect_term, return_tensors="pt", padding=True, truncation=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)
        label_idx = torch.argmax(outputs.logits, dim=1).item()

    # 💡 核心修正：0 代表 NEGATIVE, 1 代表 POSITIVE
    mapping = {0: 'NEGATIVE', 1: 'POSITIVE'}
    return mapping.get(label_idx, 'NEGATIVE')


# ==========================================
# 2. AUTOMATED EVALUATION (纯二分类评估与混淆矩阵)
# ==========================================
print("\n[Step 1] Dynamically Generating Test Set via Label-Alignment...")

# 随机抽样 50 条进行闭卷测试
eval_sample_df = df_test.sample(1000, random_state=42)

y_true = []
y_pred = []

for _, row in eval_sample_df.iterrows():
    doc_sentiment = 'POSITIVE' if row['label'] == 1 else 'NEGATIVE'
    review_text = row['review']

    doc = nlp(review_text.lower())

    found_test_case = False
    for sent in doc.sents:
        for chunk in sent.noun_chunks: # spaCy 名词短语提取
            term = chunk.text.strip()

            for category, keywords in movie_aspects.items():
                if any(keyword in term for keyword in keywords):
                    pred_sentiment = predict_aspect_sentiment(sent.text, term)

                    y_true.append(doc_sentiment)
                    y_pred.append(pred_sentiment)
                    found_test_case = True
                    break
            if found_test_case: break
        if found_test_case: break

print("\n============= AUTOMATED ABSA MODEL PERFORMANCE REPORT =============")
if len(y_true) > 0:
    print(classification_report(y_true, y_pred, target_names=['NEGATIVE', 'POSITIVE'], zero_division=0))
    print(f"Empirical Alignment Accuracy: {accuracy_score(y_true, y_pred):.4f}")
    print("====================================================================")

    # 绘制纯二分类混淆矩阵
    cm = confusion_matrix(y_true, y_pred, labels=['NEGATIVE', 'POSITIVE'])
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Predicted NEGATIVE', 'Predicted POSITIVE'],
                yticklabels=['Actual NEGATIVE', 'Actual POSITIVE'])
    plt.title('ABSA Automated Alignment: Confusion Matrix', fontsize=12, fontweight='bold')
    plt.ylabel('Ground Truth (Document Label)')
    plt.xlabel('Transformer Predictions (Aspect Level)')
    plt.tight_layout()
    plt.show()
else:
    print("No strong aspect signals alignment found in sample subset.")
    print("====================================================================")


# ==========================================
# 3. MAIN DATASET PROCESSING (大规模测试集纯二分类统计)
# ==========================================
print("\n[Step 2] Processing Main Test Subset For Profiling Graph Predictions...")

aspect_sentiments = {aspect: [] for aspect in movie_aspects}

# 采样 100 条数据来跑数
main_sample_df = df_test.sample(100, random_state=100)

for review in main_sample_df['review']:
    doc = nlp(review.lower())

    for sent in doc.sents:
        for chunk in sent.noun_chunks:
            term = chunk.text.strip()

            matched_category = None
            for category, keywords in movie_aspects.items():
                if any(keyword in term for keyword in keywords):
                    matched_category = category
                    break

            if matched_category:
                sentiment = predict_aspect_sentiment(sent.text, term)
                aspect_sentiments[matched_category].append(sentiment)
                break


# ==========================================
# 4. RESULTS DISPLAY & VISUALIZATION (去除了中性的干净大图)
# ==========================================
print("\n[Step 3] Rendering Final Distribution Mapping Matrix...")

print(f"\n{'Aspect':<15} | {'Positive':>10} | {'Negative':>10} | {'Total':>8}")
print("-" * 50)
for aspect, sentiments in aspect_sentiments.items():
    pos = sentiments.count('POSITIVE')
    neg = sentiments.count('NEGATIVE')
    total = len(sentiments)
    print(f"{aspect:<15} | {pos:>10} | {neg:>10} | {total:>8}")

aspect_names = list(aspect_sentiments.keys())
pos_counts = [aspect_sentiments[a].count('POSITIVE') for a in aspect_names]
neg_counts = [aspect_sentiments[a].count('NEGATIVE') for a in aspect_names]

x = np.arange(len(aspect_names))
width = 0.35 # 两个柱子，宽度可以稍微放宽一点点

plt.figure(figsize=(11, 6))
plt.bar(x - width/2, pos_counts, width, label='Positive', color='steelblue')
plt.bar(x + width/2, neg_counts, width, label='Negative', color='salmon')

plt.xticks(x, aspect_names)
plt.title('DistilBERT-ABSA: Aspect-wise Sentiment Maps by Movie Aspect (100 Sample Rows)', fontsize=12, fontweight='bold')
plt.ylabel('Observed Count of Mentions', fontsize=10)
plt.xlabel('Extracted Movie Aspect Categories', fontsize=10)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
